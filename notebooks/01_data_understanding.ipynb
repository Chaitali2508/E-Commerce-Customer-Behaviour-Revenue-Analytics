```python
orders_sorted = orders.sort_values(["customer_id", "order_purchase_timestamp"])
orders_sorted[["customer_id", "order_purchase_timestamp"]].head(10)
```

```python
overall_avg_gap = avg_gap_per_customer["avg_days_between_orders"].mean()
print(f"Average time between orders (repeat customers): {overall_avg_gap}")
```

```python
median_gap = avg_gap_per_customer["avg_days_between_orders"].median()
print(f"Median time between orders (repeat customers): {median_gap}")
```

```python
snapshot_date = orders["order_purchase_timestamp"].max() + pd.Timedelta(days=1)
print(snapshot_date)
```

```python
last_order = orders.groupby("customer_id")["order_purchase_timestamp"].max().reset_index()
last_order = last_order.rename(columns={"order_purchase_timestamp": "last_order_date"})
last_order["recency_days"] = (snapshot_date - last_order["last_order_date"]).dt.days
last_order.head(10)
```

```python
rfm = customer_orders.merge(last_order[["customer_id", "recency_days"]], on="customer_id", how="left")
rfm = rfm.rename(columns={"order_count": "frequency", "total_revenue": "monetary"})
rfm[["customer_id", "recency_days", "frequency", "monetary"]].head(10)
```

```python
rfm["R_score"] = pd.qcut(rfm["recency_days"], 5, labels=[5, 4, 3, 2, 1])
rfm["F_score"] = pd.qcut(rfm["frequency"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5])
rfm["M_score"] = pd.qcut(rfm["monetary"], 5, labels=[1, 2, 3, 4, 5])

rfm[["customer_id", "recency_days", "R_score", "frequency", "F_score", "monetary", "M_score"]].head(10)
```

```python
rfm["RFM_score"] = rfm["R_score"].astype(str) + rfm["F_score"].astype(str) + rfm["M_score"].astype(str)
rfm[["customer_id", "R_score", "F_score", "M_score", "RFM_score"]].head(10)
```

```python
def segment_customer(row):
    r = int(row["R_score"])
    fm = (int(row["F_score"]) + int(row["M_score"])) / 2

    if r >= 4 and fm >= 4:
        return "Champions"
    elif r >= 3 and fm >= 3:
        return "Loyal Customers"
    elif r >= 4 and fm < 3:
        return "New/Promising"
    elif r <= 2 and fm >= 4:
        return "At Risk"
    elif r <= 2 and fm <= 2:
        return "Lost"
    else:
        return "Needs Attention"

rfm["segment"] = rfm.apply(segment_customer, axis=1)
rfm["segment"].value_counts()
```

```python
category_performance = master.groupby("product_category_name").agg(
    total_revenue=("order_total_value", "sum"),
    order_count=("order_id", "nunique")
).reset_index().sort_values("total_revenue", ascending=False)

category_performance.head(10)
```

```python
top_categories = category_performance.head(10)

plt.figure(figsize=(10, 6))
plt.barh(top_categories["product_category_name"], top_categories["total_revenue"])
plt.gca().invert_yaxis()  # highest revenue at top
plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Total Revenue")
plt.tight_layout()
plt.show()
```
