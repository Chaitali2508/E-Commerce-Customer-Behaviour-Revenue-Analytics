```python
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_approved_at"] = pd.to_datetime(orders["order_approved_at"])
```

```python
orders["approval_delay"] = orders["order_approved_at"] - orders["order_purchase_timestamp"]
```

```python
avg_delay = orders["approval_delay"].mean()
print(avg_delay)
```

```python
orders["approval_delay"] = orders["order_approved_at"] - orders["order_purchase_timestamp"]
orders["approval_delay"].isnull().sum()
```

```python
products_clean = products.copy()
products_clean["product_category_name"] = products_clean["product_category_name"].fillna("unknown")
```

```python
median_weight = products_clean["product_weight_g"].median()
products_clean["product_weight_g"] = products_clean["product_weight_g"].fillna(median_weight)
```

```python
master = orders.merge(customers_clean, on="customer_id", how="left")
print(master.shape)
master.head()
```

```python
master = master.merge(order_items, on="order_id", how="left")
print(master.shape)
```

```python
master = master.merge(products_clean, on="product_id", how="left")
print(master.shape)
```

```python
master = master.merge(payments, on="order_id", how="left")
print(master.shape)
```

```python
master["order_month"] = master["order_purchase_timestamp"].dt.to_period("M")
master[["order_purchase_timestamp", "order_month"]].head()
```

```python
master["order_total_value"] = master["price"] + master["shipping_charges"]
master[["price", "shipping_charges", "order_total_value"]].head()
```

```python
customer_revenue = master.groupby("customer_id")["order_total_value"].sum().reset_index()
customer_revenue = customer_revenue.rename(columns={"order_total_value": "total_revenue"})
customer_revenue.sort_values("total_revenue", ascending=False).head(10)
```

```python
customer_orders = master.groupby("customer_id").agg(
    total_revenue=("order_total_value", "sum"),
    order_count=("order_id", "nunique"),
    avg_order_value=("order_total_value", "mean")
).reset_index()

customer_orders.sort_values("order_count", ascending=False).head(10)
```

```python
customer_orders["order_count"].describe()
```

```python
customer_orders["order_count"].value_counts().sort_index()
```

```python
customer_orders["is_repeat"] = customer_orders["order_count"] > 1

repeat_rate = customer_orders["is_repeat"].mean() * 100
print(f"Repeat customer rate: {repeat_rate:.2f}%")

customer_orders["is_repeat"].value_counts()
```

```python
orders_sorted = orders.sort_values(["customer_id", "order_purchase_timestamp"])
orders_sorted[["customer_id", "order_purchase_timestamp"]].head(10)
```

```python
orders_sorted["days_since_last_order"] = orders_sorted.groupby("customer_id")["order_purchase_timestamp"].diff()
orders_sorted[["customer_id", "order_purchase_timestamp", "days_since_last_order"]].head(10)
```

```python
avg_gap_per_customer = orders_sorted.groupby("customer_id")["days_since_last_order"].mean().reset_index()
avg_gap_per_customer = avg_gap_per_customer.rename(columns={"days_since_last_order": "avg_days_between_orders"})
avg_gap_per_customer.head(10)
```

```python
overall_avg_gap = avg_gap_per_customer["avg_days_between_orders"].mean()
print(f"Average time between orders (repeat customers): {overall_avg_gap}")
```

```python
median_gap = avg_gap_per_customer["avg_days_between_orders"].median()
print(f"Median time between orders (repeat customers): {median_gap}")
```
